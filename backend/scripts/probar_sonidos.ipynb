# Reproducir sonidos del reloj ZKTeco



```python
from zk import ZK
from contextlib import suppress
import time

ZK_IP = "192.168.137.2"
ZK_PORT = 4370
ZK_PASSWORD = 171
ZK_TIMEOUT = 5


def connect_zk():
    zk = ZK(
        ZK_IP,
        port=ZK_PORT,
        timeout=ZK_TIMEOUT,
        password=ZK_PASSWORD,
        force_udp=False,
        ommit_ping=False,
    )
    return zk.connect()
```

## Tabla de índices de voz


```python
VOICES = {
    0: "Thank You",
    1: "Incorrect Password",
    2: "Access Denied",
    3: "Invalid ID",
    4: "Please try again",
    5: "Duplicate ID",
    6: "The clock is flow",
    7: "The clock is full",
    8: "Duplicate finger",
    9: "Duplicated punch",
    10: "Beep kuko",
    11: "Beep siren",
    13: "Beep bell",
    18: "Windows(R) opening sound",
    20: "Fingerprint not emolt",
    21: "Password not emolt",
    22: "Badges not emolt",
    23: "Face not emolt",
    24: "Beep standard",
    30: "Invalid user",
    31: "Invalid time period",
    32: "Invalid combination",
    33: "Illegal Access",
    34: "Disk space full",
    35: "Duplicate fingerprint",
    36: "Fingerprint not registered",
    51: "Focus eyes on the green box",
}

for index, label in VOICES.items():
    print(f"{index:>3} - {label}")
```

## Reproducir un solo sonido

```python
VOICE_INDEX = 10

conn = None
device_disabled = False

try:
    print("Conectando")
    conn = connect_zk()
    print("exitosa.")

    conn.disable_device()
    device_disabled = True

    label = VOICES.get(VOICE_INDEX, "desconocido")
    print(f"Reproduciendo sonido {VOICE_INDEX} ({label})...")
    conn.test_voice(index=VOICE_INDEX)

except Exception as e:
    print(f"Error: {e}")

finally:
    if conn:
        if device_disabled:
            with suppress(Exception):
                conn.enable_device()
        with suppress(Exception):
            conn.disconnect()
    print("cerrada.")
```

```python
DELAY_SECONDS = 2.5

conn = None
device_disabled = False

try:
    print("Conectando al reloj...")
    conn = connect_zk()
    print("Conexion exitosa.")

    conn.disable_device()
    device_disabled = True

    for index, label in VOICES.items():
        try:
            print(f"[{index:>3}] {label}")
            conn.test_voice(index=index)
        except Exception as e:
            print(f"  -> Error reproduciendo indice {index}: {e}")
        time.sleep(DELAY_SECONDS)

except Exception as e:
    print(f"Error: {e}")

finally:
    if conn:
        if device_disabled:
            with suppress(Exception):
                conn.enable_device()
        with suppress(Exception):
            conn.disconnect()
    print("Conexion cerrada.")
```
